# Data Cleaning & Preparation

## Purpose 
The purpose of this notebook is to prepare the labor force and unemployment dataset for analysis. It will be making consistent column names, filtering dataset to a set geographic scope, and creating a clean dataset for visualization and trend analysis

```python
#Importing Data 
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/laborforceandunemployment_annual_2025421.csv")
```

```python
#Cleaning Column Names 
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)
```

```python
#Checking
df.columns
```

```python
#County Level Filter
df_county = df[df['area_type']== 'County']
```

```python
#Shape
df_county.shape
```

```python
#Checking
df_county['area_name'].nunique()
```

```python
df_county['area_type'].unique()
```

```python
#Sorting
df_county = df_county.sort_values(by=['area_name', 'year'])
```

```python
#Importing into cleaned data folder
output_path = "../data/cleaned/california_county_unemployment_cleaned.csv"
df_county.to_csv(output_path, index = False)
```
